```python
# Импортируйте все необходимые библиотеки и модули которые вам необходимы
import pandas as pd
```

```python
# Загрузите датасет и проведите его очистку и форматирование если это необходимо

df = pd.read_csv('baseball_games.csv', index_col=0)
for i, row in df[df.isna().any(axis=1)].iterrows():
    print(f'строка {i}: нет значения в поле {row.index[row.isna()].tolist()}')
df['date'] = pd.to_datetime(df['date'])
df['wind_direction'] = df['wind_direction'].str.strip()
df['temperature'] = ((df['temperature'] - 32) * 5 / 9).round(1)
```

# 13. Выведите количество игр которая сыграла каждая команда в данном сезоне?

```python
reg_season_games = df[df['season'] == 'regular season']
total_games = reg_season_games['home_team'].value_counts() + reg_season_games['away_team'].value_counts()
total_games.sort_values(ascending=False)
```

Здесь и далее в задачах с условием "в сезоне" принимаем, что имеется в виду групповой этап (regular season). для этого вначале сфильтруем по типу сезона.

# 14. Какая команда выиграла наибольшое количество матчей в сезоне? (Будьте внимательны с типом матча в сезоне - season).

```python
home_wins = reg_season_games[reg_season_games['home_team_win'] == 1]['home_team'].value_counts()
away_wins = reg_season_games[reg_season_games['home_team_loss'] == 1]['away_team'].value_counts()
total_wins = (home_wins + away_wins).fillna(0)
print('Самая результативная команда в сезоне:', total_wins.idxmax(), '-', total_wins.max(), 'побед')
```

поскольку параметр "home_team_outcome" определяет победу или поражение для домашней команды матча, победы гостей определяем по поражению хозяев. перестраховываемся и заменяем при сложении сериез пустые значения нулями для случая, если бы у кого-то не было домашних или гостевых побед.
